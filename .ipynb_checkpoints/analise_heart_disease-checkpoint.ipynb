## 1. Introdução

Este notebook contém a análise do dataset 'Heart Disease' do repositório UCI. O objetivo é aplicar técnicas de aprendizado de máquina para resolver dois problemas:

1.  **Classificação:** Prever a presença ou ausência de doença cardíaca em um paciente com base em seus dados clínicos.
2.  **Clusterização:** Agrupar pacientes com características semelhantes para identificar perfis de risco.

## 2. Carregamento e Descrição dos Dados

Primeiro, vamos importar as bibliotecas necessárias e carregar o dataset usando a biblioteca `ucimlrepo`.

```python
import pandas as pd
import numpy as np
from ucimlrepo import fetch_ucirepo

# Buscar o dataset pelo seu ID no repositório UCI
heart_disease = fetch_ucirepo(id=45)

# Separar os dados em features (X) e target (y)
X = heart_disease.data.features
y = heart_disease.data.targets

# O target 'num' indica a presença de doença (valores > 0)
# Vamos transformar em um problema binário: 0 (ausência) vs 1 (presença)
y_binary = (y > 0).astype(int)

print("Dimensões das Features (X):", X.shape)
print("Dimensões do Target (y):", y_binary.shape)

print("\n--- Primeiras 5 linhas das Features (X): ---")
display(X.head())
```
